```python
# Ultra-Fast Hybrid CNN + ViT (MobileNetV2 + EfficientNetB0)

import os
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.applications import MobileNetV2, EfficientNetB0
from tensorflow.keras.layers import Input, Flatten, Dense, Concatenate, Dropout, Reshape
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.image import ImageDataGenerator

from sklearn.metrics import confusion_matrix, classification_report


# ==============================
# USE ALL CPU CORES
# ==============================

os.environ["OMP_NUM_THREADS"] = "8"
os.environ["TF_NUM_INTRAOP_THREADS"] = "8"
os.environ["TF_NUM_INTEROP_THREADS"] = "8"

tf.config.threading.set_intra_op_parallelism_threads(8)
tf.config.threading.set_inter_op_parallelism_threads(8)


# ==============================
# PARAMETERS
# ==============================

IMAGE_SIZE = 96
BATCH_SIZE = 24
EPOCHS = 25
LEARNING_RATE = 3e-4

PATCH_SIZE = 8
PROJECTION_DIM = 64
NUM_HEADS = 2
MLP_UNITS = 128

DATASET_PATH = "/content/PlantVillage"


# ==============================
# DATA LOADING
# ==============================

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=15,
    zoom_range=0.15,
    horizontal_flip=True,
    validation_split=0.15
)

test_datagen = ImageDataGenerator(rescale=1./255)


train_gen = train_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=(IMAGE_SIZE, IMAGE_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    subset="training"
)

val_gen = train_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=(IMAGE_SIZE, IMAGE_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    subset="validation"
)

test_gen = test_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=(IMAGE_SIZE, IMAGE_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)


NUM_CLASSES = train_gen.num_classes
print("Number of Classes:", NUM_CLASSES)


# ==============================
# CNN FEATURE EXTRACTORS (LIGHT)
# ==============================

def build_cnn_models(input_tensor):

    mob = MobileNetV2(
        weights="imagenet",
        include_top=False,
        input_tensor=input_tensor
    )

    eff = EfficientNetB0(
        weights="imagenet",
        include_top=False,
        input_tensor=input_tensor
    )


    # Fine-tune last 20 layers
    for model in [mob, eff]:

        for layer in model.layers[:-20]:
            layer.trainable = False

        for layer in model.layers[-20:]:
            layer.trainable = True


    mob_f = Flatten()(mob.output)
    eff_f = Flatten()(eff.output)

    merged = Concatenate()([mob_f, eff_f])

    return merged


# ==============================
# ViT BLOCK (LIGHT)
# ==============================

def build_vit(num_patches):

    inputs = Input(shape=(num_patches, PROJECTION_DIM))
    x = inputs


    # Only ONE Transformer block
    x1 = tf.keras.layers.LayerNormalization(epsilon=1e-6)(x)

    attention = tf.keras.layers.MultiHeadAttention(
        num_heads=NUM_HEADS,
        key_dim=PROJECTION_DIM,
        dropout=0.01
    )(x1, x1)

    x2 = tf.keras.layers.Add()([attention, x])


    x3 = tf.keras.layers.LayerNormalization(epsilon=1e-6)(x2)

    x3 = Dense(MLP_UNITS, activation="relu")(x3)
    x3 = Dense(PROJECTION_DIM)(x3)
    x3 = Dropout(0.01)(x3)


    x = tf.keras.layers.Add()([x3, x2])


    x = Flatten()(x)
    x = Dropout(0.4)(x)

    outputs = Dense(NUM_CLASSES, activation="softmax")(x)

    return Model(inputs, outputs)


# ==============================
# BUILD MODEL
# ==============================

input_img = Input(shape=(IMAGE_SIZE, IMAGE_SIZE, 3))


cnn_features = build_cnn_models(input_img)

cnn_features = Dense(256, activation="relu")(cnn_features)


num_patches = (IMAGE_SIZE // PATCH_SIZE) ** 2

projected = Dense(num_patches * PROJECTION_DIM)(cnn_features)

patch_ready = Reshape((num_patches, PROJECTION_DIM))(projected)


vit_model = build_vit(num_patches)

outputs = vit_model(patch_ready)


model = Model(input_img, outputs)


# ==============================
# COMPILE
# ==============================

model.compile(

    optimizer=tf.keras.optimizers.Adam(learning_rate=LEARNING_RATE),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)

model.summary()


# ==============================
# TRAINING
# ==============================

callbacks = [

    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=6,
        restore_best_weights=True
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=3,
        min_lr=1e-6,
        verbose=1
    )
]


history = model.fit(

    train_gen,

    epochs=EPOCHS,

    validation_data=val_gen,

    steps_per_epoch=len(train_gen),
    validation_steps=len(val_gen),

    callbacks=callbacks
)


# ==============================
# PLOT
# ==============================

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.plot(history.history["accuracy"], label="Train")
plt.plot(history.history["val_accuracy"], label="Validation")
plt.title("Accuracy")
plt.legend()

plt.subplot(1,2,2)
plt.plot(history.history["loss"], label="Train")
plt.plot(history.history["val_loss"], label="Validation")
plt.title("Loss")
plt.legend()

plt.show()


# ==============================
# TEST
# ==============================

loss, acc = model.evaluate(test_gen)

print("Test Accuracy:", acc)


# ==============================
# CONFUSION MATRIX
# ==============================

y_true = test_gen.classes

y_pred = model.predict(test_gen)

y_pred = np.argmax(y_pred, axis=1)


cm = confusion_matrix(y_true, y_pred)

print("Confusion Matrix")
print(cm)


print("Classification Report")

print(classification_report(
    y_true,
    y_pred,
    target_names=list(test_gen.class_indices.keys())
))


# ==============================
# SAVE
# ==============================

model.save("hybrid_mobilenet_effnet_vit_fast.keras")

print("Model Saved Successfully")
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.preprocessing.image import ImageDataGenerator

from sklearn.metrics import (
    roc_curve, auc,
    confusion_matrix,
    cohen_kappa_score,
    precision_score,
    recall_score,
    f1_score,
    accuracy_score
)

from sklearn.preprocessing import label_binarize


# ==============================
# LOAD MODEL
# ==============================

model = tf.keras.models.load_model("/content/hybrid_mobilenet_effnet_vit_fast.keras")


# ==============================
# DATASET SETTINGS
# ==============================

DATASET_PATH ="/content/PlantVillage"

IMAGE_SIZE = 96
BATCH_SIZE = 24


# ==============================
# LOAD TEST DATA
# ==============================

test_datagen = ImageDataGenerator(rescale=1./255)

test_gen = test_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=(IMAGE_SIZE, IMAGE_SIZE),
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)


NUM_CLASSES = test_gen.num_classes
CLASS_NAMES = list(test_gen.class_indices.keys())


# ==============================
# PREDICTIONS
# ==============================

y_true = test_gen.classes

y_prob = model.predict(test_gen)

y_pred = np.argmax(y_prob, axis=1)


# ==============================
# CONFUSION MATRIX
# ==============================

cm = confusion_matrix(y_true, y_pred)


# ==============================
# CALCULATE METRICS
# ==============================

# Accuracy
accuracy = accuracy_score(y_true, y_pred)

# Cohen’s Kappa
kappa = cohen_kappa_score(y_true, y_pred)

# Precision (Macro)
precision = precision_score(y_true, y_pred, average="macro")

# Recall (Macro)
recall = recall_score(y_true, y_pred, average="macro")

# F1-Score (Macro)
f1 = f1_score(y_true, y_pred, average="macro")


# ==============================
# SPECIFICITY (MULTI-CLASS)
# ==============================

specificity_list = []

for i in range(len(cm)):

    tn = np.sum(np.delete(np.delete(cm, i, axis=0), i, axis=1))
    fp = np.sum(cm[:, i]) - cm[i, i]

    spec = tn / (tn + fp)
    specificity_list.append(spec)

specificity_mean = np.mean(specificity_list)


# ==============================
# PRINT METRICS
# ==============================

print("\n===== MODEL PERFORMANCE METRICS =====\n")

print(f"Accuracy      : {accuracy:.4f}")
print(f"Cohen's Kappa : {kappa:.4f}")
print(f"Precision     : {precision:.4f}")
print(f"Recall        : {recall:.4f}")
print(f"F1-Score      : {f1:.4f}")
print(f"Specificity   : {specificity_mean:.4f}")


# ==============================
# ROC CURVE (MULTI-CLASS)
# ==============================

# Binarize labels
y_true_bin = label_binarize(y_true, classes=range(NUM_CLASSES))


# Color map
colors = plt.cm.tab20(np.linspace(0, 1, NUM_CLASSES))


plt.figure(figsize=(13,10))


for i, color in zip(range(NUM_CLASSES), colors):

    fpr, tpr, _ = roc_curve(y_true_bin[:, i], y_prob[:, i])
    roc_auc = auc(fpr, tpr)

    plt.plot(
        fpr,
        tpr,
        color=color,
        linewidth=2,
        label=f"{CLASS_NAMES[i]} (AUC={roc_auc:.2f})"
    )


# Reference line
plt.plot([0,1], [0,1], "k--", linewidth=1.5)


# Labels
plt.xlabel("False Positive Rate", fontsize=13)
plt.ylabel("True Positive Rate", fontsize=13)

plt.title("Multi-Class ROC Curve (All Classes)", fontsize=15)


# Legend outside
plt.legend(
    loc="center left",
    bbox_to_anchor=(1.02, 0.5),
    fontsize=9
)


plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import tensorflow as tf
import numpy as np
import cv2
import os
import matplotlib.pyplot as plt
from google.colab import files

# Load trained model
model = tf.keras.models.load_model("/content/hybrid_mobilenet_effnet_vit_fast.keras")
print("Model loaded successfully")

# Dataset path (for class names)
dataset_path = "/content/PlantVillage"
class_names = sorted(os.listdir(dataset_path))

# Get model input size
IMAGE_SIZE = model.input_shape[1]

# Upload image
uploaded = files.upload()

# Get uploaded image name
image_path = list(uploaded.keys())[0]

# Read image
img = cv2.imread(image_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Resize image
img_resized = cv2.resize(img, (IMAGE_SIZE, IMAGE_SIZE))

# Normalize
img_resized = img_resized / 255.0

# Expand dimension
img_input = np.expand_dims(img_resized, axis=0)

# Prediction
prediction = model.predict(img_input)

predicted_class = np.argmax(prediction)
confidence = np.max(prediction) * 100

print("Predicted Disease:", class_names[predicted_class])
print("Confidence Score:", round(confidence,2), "%")

# Show image
plt.imshow(img)
plt.title(f"{class_names[predicted_class]} ({confidence:.2f}%)")
plt.axis("off")
plt.show()
```
